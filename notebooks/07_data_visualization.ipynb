```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r".../datasets/Ice Cream Ratings.csv")
df = df.set_index("Date")
df
```

```python
print(plt.style.available)
plt.style.use("fivethirtyeight")
```

```python
df.plot(legend=True) # To create a line plot, you can use the plot() method on the dataframe. This will create a line plot for each column in the dataframe, showing the trends over time or across categories. You can also customize the plot by adding labels, titles, and other parameters.
```

```python
df.plot(kind = "line", subplots = True)
```

```python
df.plot(kind = "line", title = "Ice Cream Ratings", xlabel = "Daily Ratings", ylabel = "Scores") # You can also customize the plot by adding a title, labels for the x and y axes, and other parameters. For example, you can specify the kind of plot you want to create (e.g., line, bar, scatter) and add a legend to differentiate between the different columns in the dataframe.
```

```python
df.plot(kind = "bar")
```

```python
df.plot(kind = "bar", stacked = True) # If you want to create a stacked bar plot, you can specify stacked=True. This will stack the bars on top of each other, allowing you to see the total values for each category as well as the individual contributions from each column.
```

```python
df["Flavor Rating"].plot(kind = "bar", stacked = True)
```

```python
df.plot(kind = "barh", stacked = True)
```

```python
df.plot.scatter(x = "Texture Rating", y = "Overall Rating", s = 500 , c = "red") # To create a scatter plot, you can use the plot.scatter() method and specify the x and y columns you want to plot. You can also customize the size of the markers using the s parameter.
```

```python
df.plot.hist(bins = 10) # To create a histogram, you can use the plot.hist() method and specify the number of bins you want to use. This will create a histogram for each column in the dataframe, showing the distribution of values for each column.
```

```python
df.boxplot() # To create a box plot, you can use the plot.box() method. This will create a box plot for each column in the dataframe, showing the distribution of values and any potential outliers.
```

```python
df.plot.area(figsize = (10, 6)) # To create an area plot, you can use the plot.area() method. This will create an area plot for each column in the dataframe, showing the cumulative values for each column over time or across categories.
```

```python
df.plot.pie(y = "Flavor Rating") # To create a pie chart, you can use the plot.pie() method and specify the column you want to plot using the y parameter. This will create a pie chart showing the proportion of each category in the specified column. You can also customize the pie chart by adding labels, colors, and other parameters.
```
