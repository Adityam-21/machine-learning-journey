```python
# Broadcasting

import numpy as np

array_1 = np.array([[0,0,0],
                   [10,10,10],
                   [20,20,20],
                   [30,30,30]])

array_2 = np.array([1,2,3])

print(array_1.shape) #(4,3)
print(array_2.shape) #( ,3) -> dimensions match plus there is a '1' . Hence cond. satisfied

#array_1 + array_2
```

```python
# Filtering

ages = np.array([[21,17,19,20,16,30,18,65],
                 [39,22,15,99,18,19,20,21]])

adults = np.where(ages >= 18 , ages , 0) # only use 'where' clause when you need to preserve the above data otherwise use boolean filtering as its much faster

print(adults)
```
