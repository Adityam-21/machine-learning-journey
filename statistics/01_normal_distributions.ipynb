```python
df1 = sns.load_dataset('iris')
df1.head()
```

```python
sns.histplot(df1['sepal_length'], kde=True) # dosen't follow gaussian distribution
```

```python
sns.histplot(df1['sepal_width'], kde=True) # follow gaussian distribution

# Therefore, we can apply 68-95-99.7 rule to sepal width but not to sepal length.
```

```python
sns.countplot(x=df1['species']) # countplot is used to count the number of occurrences of each category in a categorical variable. In this case, it counts the number of occurrences of each species in the 'species' column of the iris dataset.
```

```python
np.percentile(df1['sepal_length'], [25,75])

# Hence, 6.4 - 5.1 = 1.3 is the IQR of sepal length. We can also cal. the IQR of sepal width by 75-25 = 50th percentile
```
