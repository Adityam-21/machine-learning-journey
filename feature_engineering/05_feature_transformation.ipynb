```python
import pandas as pd
import numpy as np
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = pd.read_csv(r"C:\Users\kadit\ML-Journey\datasets\covid_toy.csv")
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.drop(columns=['has_covid']), df['has_covid'], test_size=0.2)
```

```python
X_train.head()
```

1. Without Column Transformer:

```python
# Adding simpleimputer to 'fever'col.

si = SimpleImputer ()
X_train_fever = si.fit_transform(X_train[['fever']])

# Also the test data

X_test_fever = si.transform(X_test[['fever']])

X_train_fever.shape, X_test_fever.shape
```

```python
# OE -> 'cough' col.

oe = OrdinalEncoder(categories=[['Mild', 'Strong']])
X_train_cough = oe.fit_transform(X_train[['cough']])

# Also the test data

X_test_cough = oe.transform(X_test[['cough']])

X_train_cough.shape, X_test_cough.shape
```

```python
# OHE on 'gender' and 'city' col.

ohe = OneHotEncoder(drop ='first')
X_train_gender_city = ohe.fit_transform(X_train[['gender', 'city']])

# Also the test data

X_test_gender_city = ohe.transform(X_test[['gender', 'city']])

X_train_gender_city.shape
```

```python
# Extracting 'age' col. 

X_train_age = X_train.drop (columns=['gender', 'city', 'fever', 'cough']).values

# Also the test data

X_test_age = X_test.drop (columns=['gender', 'city', 'fever', 'cough']).values

X_train_age.shape, X_test_age.shape
```

```python
# Concatenating all the transformed features into a single array for training and testing:

X_train_transformed = np.concatenate([X_train_fever, X_train_cough, X_train_gender_city.toarray(), X_train_age], axis=1)

X_test_transformed = np.concatenate([X_test_fever, X_test_cough, X_test_gender_city.toarray(), X_test_age], axis=1)

X_train_transformed.shape, X_test_transformed.shape
```

2. With Column Transform:

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer(transformers = [
    ('tnf1', SimpleImputer (), ['fever']),
    ('tnf2', OrdinalEncoder(categories=[['Mild', 'Strong']]), ['cough']),
    ('tnf3', OneHotEncoder(drop ='first'), ['gender', 'city'])
    ], remainder = 'passthrough')
```

```python
X_train_transformed = transformer.fit_transform(X_train)
X_test_transformed = transformer.transform(X_test)
```
